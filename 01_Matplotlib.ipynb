```python
import matplotlib.pyplot as plt
```

```python
# bar chart
categories = ['A', 'B', 'C', 'D', 'E']
values = [10, 24, 38, 18, 27]

plt.figure(figsize = (6, 4))
plt.bar(categories, values, color = "brown", edgecolor = "black")
plt.title("Bar chart", fontsize = 20)
plt.xlabel("Categories", fontsize = 16)
plt.ylabel("Values", fontsize = 16)
plt.tight_layout()
plt.show()
```

```python
# Line Chart
months = ['Jan', 'Feb', 'March', 'Apr', 'May', 'June']
sales = [150, 200, 190, 220, 300, 250]

plt.figure(figsize=(10, 4))
plt.plot(months, sales, marker = "o", linewidth = 3, markersize = 10, markerfacecolor = "black", markeredgecolor = "black")
plt.title("line Chart", fontsize = 20)
plt.xlabel("Months", fontsize = 16)
plt.ylabel("Sales", fontsize = 16)
plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
# Area Chart
months = ['Jan', 'Feb', 'March', 'Apr', 'May', 'June']
sales = [150, 200, 190, 220, 300, 250]

plt.figure(figsize=(10, 4))
plt.fill_between(months, sales, color = "skyblue", alpha = 0.5)
plt.plot(months, sales, marker = "o", linewidth = 3, markersize = 10, markerfacecolor = "black", markeredgecolor = "black")
plt.title("Area Chart", fontsize = 20)
plt.xlabel("Months", fontsize = 16)
plt.ylabel("Sales", fontsize = 16)
plt.tight_layout()
plt.grid(False)
plt.show()
```

```python
# Pie chart
categories = ['Apple','Bananas', 'Cherries', 'Grapes']
values = [25, 35, 20, 20]

plt.figure(figsize = (10, 4))
plt.pie(values, labels = categories, autopct = '%1.1f%%')
plt.title("Pie chart distribution", fontsize = 20)
plt.show()
```

```python
# Donut chart
categories = ['Apple','Bananas', 'Cherries', 'Grapes']
values = [25, 35, 20, 20]

plt.figure(figsize = (10, 4))
plt.pie(values, labels = categories, autopct = '%1.1f%%')
centre_circle = plt.Circle((0, 0), 0.40, fc = "white")
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.title("Pie chart distribution", fontsize = 20)
plt.show()
```

```python
# Histogram
import numpy as np
data = np.random.normal(50, 10, 200)
plt.figure(figsize = (10, 4))
plt.hist(data, bins =10, color = "orange", edgecolor = "black")
plt.title("Histogram", fontsize = 20)
plt.xlabel("Values", fontsize = 16)
plt.ylabel("Frequencies", fontsize = 16)
plt.tight_layout()
plt.show()
```

```python
# Box plot
plt.figure(figsize = (10, 4))
sns.boxplot(x = 'day', y = 'total_bill', data = data, palette = 'pastel')
plt.title('Box plot disttribution', fontsize = 20)
plt.show()
# these outside values -> outliers
# box middle line -> median(q2) - 50%
# box upper line -> max value - quantile(q3) - 75%
# box lower line -> min value - quantile(q1) - 25%
```

```python
# Violin Plot - show destribution along the median
plt.figure(figsize = (10, 4))
sns.violinplot(x = 'day', y = 'total_bill', data = data, palette = 'Set1')
plt.title('Violin plot disttribution', fontsize = 20)
plt.show()
```

```python
# Pair plot: multiple scatter plots for feature relationship
import seaborn as sns
data = sns.load_dataset('iris') # flower breed dataset
data.head()
```

```python
plt.figure(figsize = (10, 4))
sns.pairplot(data, hue = 'species', diag_kind = 'kde', palette = 'Set1')
plt.show()

# it shows:
# data distribution
# correlation between freatures
# clusters (groups)
# sepration between classes
```

```python
data['species'].unique()
```

```python
# Heat Map - cirrelation heatmap (numeric values)
# positive correlation - increases
# negative correlation - decreses
# No correlation - (may increases or decreases or constant etc)

corr = data.corr(numeric_only = True)
corr
```

```python
# positive - right prediction for class 1
# negative - not right prediction for class 1
plt.figure(figsize = (10, 4))
sns.heatmap(corr, annot = True, cmap = 'coolwarm', linewidth = 2)
plt.show()
```

```python
# Regression Plot
data = sns.load_dataset('tips')
data.head()
```

```python
plt.figure(figsize = (10, 4))
sns.regplot(x = 'total_bill', y = 'tip', data = data)
plt.title("Regression Plot RegPlot")
plt.xlabel("Total Bill")
plt.ylabel("Tip given")
plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y = np.array([3, 4, 6, 9, 11, 56, 15, 16])

plt.figure(figsize = (10, 4))
m, c = np.polyfit(x, y, 1)
y_predict = m*x + c
plt.scatter(x, y)
plt.plot(x, y_predict)
plt.title("Regression Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
# shaded region - confidence intetval
# less wide - less accurate
# more wide - more accurate
```

```python
plt.figure(figsize = (10, 4))
sns.regplot(x = 'total_bill', y = 'tip', data = data, ci= None)
plt.title("Regression Plot RegPlot")
plt.xlabel("Total Bill")
plt.ylabel("Tip given")
plt.show()
```
