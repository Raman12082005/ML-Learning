```python
import pandas as pd
import requests
```

```python
response = requests.get("Your API KEY")
```

```python
temp_df = pd.DataFrame(response.json())
```

```python
temp_df.head()
```

```python
movies = "Batman"

df_list = []

for i in range(1, 107):
    response = requests.get(
        'Your API KEY with MODIFICATION'.format(movies, i)
    )

    data = response.json()

    if "Search" in data:
        temp_df = pd.DataFrame(data["Search"])
        df_list.append(temp_df)

df = pd.concat(df_list, ignore_index=True)
print(df)
```

```python
df.to_csv('movies.csv')
```
